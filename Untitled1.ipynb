punto 1

```python
cadena = input("Ingrese una cadena: ")
caracter = input("Ingrese un carácter: ")
i = int(input("Ingrese la posición: "))

nueva = cadena[:i] + caracter + cadena[i+1:]

print("La nueva cadena es:", nueva)
```

punto 2

```python
texto = input("Ingrese un texto: ")
caracter = input("Ingrese un carácter: ")

# a
if caracter in texto:
    print("Sí está presente")
else:
    print("No está presente")

# b
print("Primera posición:", texto.find(caracter))

# c
print("Cantidad de veces:", texto.count(caracter))
```

punto 3

```python
cadena1 = input("Ingrese la primera cadena: ")
cadena2 = input("Ingrese la segunda cadena: ")

# a) iguales
if cadena1 == cadena2:
    print("Las cadenas son iguales")
else:
    print("Las cadenas son diferentes")

# b) contenida
encontrada = False

for i in range(len(cadena1) - len(cadena2) + 1):
    if cadena1[i:i+len(cadena2)] == cadena2:
        encontrada = True

if encontrada:
    print("La segunda cadena está contenida en la primera")
else:
    print("No está contenida")
```

punto 4

```python
frase = input("Ingrese una frase: ")
nueva = ""
espacio_anterior = False
for c in frase:
    if c == " ":
        if not espacio_anterior:
            nueva += c
        espacio_anterior = True
    else:
        nueva += c
        espacio_anterior = False
print("Frase corregida:", nueva)
```

```python
def separar_palabras(oracion):
    palabra = ""
    for c in oracion:
        if c != " ":
            palabra += c
        else:
            if palabra != "":
                print(palabra)
                palabra = ""
    if palabra != "":
        print(palabra)


frase = input("Ingrese una oración: ")
separar_palabras(frase)
```
