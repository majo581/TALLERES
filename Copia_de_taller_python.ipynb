## primer punto


```python
print("VERIFICADOR DE TRIANGULOS")
lado1 = float(input("Ingrese el primer lado: "))
lado2 = float(input("Ingrese el segundo lado: "))
lado3 = float(input("Ingrese el tercer lado: "))

def verificar_triangulo(lado1, lado2, lado3):
    if lado1 + lado2 > lado3 and lado1 + lado3 > lado2 and lado2 + lado3 > lado1:
        return True
    return False

def es_triangulo_rectangulo(lado1, lado2, lado3):
    lados = sorted([lado1, lado2, lado3])
    return abs(lados[2]**2 - (lados[0]**2 + lados[1]**2)) < 0.0001

def calcular_hipotenusa(lado1, lado2, lado3):
    return max(lado1, lado2, lado3)

def tipo_triangulo(lado1, lado2, lado3):
    if lado1 == lado2 == lado3:
        return "equilatero"
    elif lado1 == lado2 or lado1 == lado3 or lado2 == lado3:
        return "isosceles"
    else:
        return "escaleno"

if verificar_triangulo(lado1, lado2, lado3):
    print("Los lados", lado1, lado2, lado3, "SI forman un triangulo")

    if es_triangulo_rectangulo(lado1, lado2, lado3):
        hipotenusa = calcular_hipotenusa(lado1, lado2, lado3)
        print("Es un triangulo rectangulo")
        print("Su hipotenusa mide:", hipotenusa)

    print("Tipo de triangulo:", tipo_triangulo(lado1, lado2, lado3))
else:
    print("Los lados", lado1, lado2, lado3, "NO forman un triangulo")
```

## segundo punto

```python
print("ANALIZADOR DE NUMEROS NATURALES ")
numero = int(input("Ingrese un numero natural mayor a 1: "))

def es_digito(n):
    if 0 <= n <= 9:
        return True
    return False

def es_par(n):
    if n % 2 == 0:
        return True
    return False

def es_perfecto(n):
    suma = 0
    for i in range(1, n):
        if n % i == 0:
            suma = suma + i
    if suma == n:
        return True
    return False

def suma_naturales(n):
    suma = 0
    for i in range(1, n + 1):
        suma = suma + i
    return suma

def es_primo(n):
    if n < 2:
        return False
    for i in range(2, n):
        if n % i == 0:
            return False
    return True

def primos_menores(n):
    lista = []
    for i in range(2, n):
        if es_primo(i):
            lista.append(i)
    return lista

def siguiente_primo(n):
    siguiente = n + 1
    while es_primo(siguiente) == False:
        siguiente = siguiente + 1
    return siguiente

if numero > 1:
    print("")
    print("--- Analisis del numero", numero, "---")

    if es_digito(numero):
        print("Es digito: SI")
    else:
        print("Es digito: NO")

    if es_par(numero):
        print("Es par: SI")
    else:
        print("Es par: NO")

    if es_perfecto(numero):
        print("Es perfecto: SI")
    else:
        print("Es perfecto: NO")

    print("Suma de 1 hasta", numero, ":", suma_naturales(numero))

    if es_primo(numero):
        print("Es primo: SI")
    else:
        print("Es primo: NO")

    print("Primos menores:", primos_menores(numero))
    print("Siguiente primo:", siguiente_primo(numero))

else:
    print("El numero debe ser mayor a 1")
```

## tercer punto

```python
print("ANALIZADOR DE ORACIONES")
oracion = input("Ingrese una oracion: ")

def contar_vocales(texto):
    vocales = "aeiouáéíóúAEIOUÁÉÍÓÚ"
    contador = 0
    for letra in texto:
        if letra in vocales:
            contador = contador + 1
    return contador

def contar_palabras(texto):
    palabras = texto.split()
    return len(palabras)

def contar_tildes(texto):
    tildes = "áéíóúÁÉÍÓÚ"
    contador = 0
    for letra in texto:
        if letra in tildes:
            contador = contador + 1
    return contador

def cambiar_vocales_a_i(texto):
    vocales = "aeiouáéíóúAEIOUÁÉÍÓÚ"
    resultado = ""
    for letra in texto:
        if letra in vocales:
            resultado = resultado + "i"
        else:
            resultado = resultado + letra
    return resultado

print("")
print("Resultados")
print("Cantidad de vocales:", contar_vocales(oracion))
print("Cantidad de palabras:", contar_palabras(oracion))
print("Cantidad de tildes:", contar_tildes(oracion))
print("Oracion cambiada:", cambiar_vocales_a_i(oracion))
```

## cuarto punto

```python

print("REPETIDOR DE CARACTER O FALSE ")

def repetir_caracter_o_false(numero, caracter, booleano):
    if booleano == True:
        return caracter * numero
    else:
        return [False] * numero

print("Ejemplo con numeros ingresados por el usuario:")
num = int(input("Ingrese un numero: "))
car = input("Ingrese un caracter: ")
opcion = input("Ingrese True o False: ")

# Esto convierte lo que escriba el usuario a True o False
if opcion.lower() == "true":
    booleano = True
else:
    booleano = False

resultado = repetir_caracter_o_false(num, car, booleano)
print("Resultado:", resultado)
```

## quinto punto


```python
lista = [15, 23, 45, 12, 30, 25, 18, 40, 35, 50]
digito = int(input("Ingrese un digito (0-9): "))

print(f"\nLista: {lista}")
print(f"Digito: {digito}\n")

def menores_al_digito(lista, digito):
    contador = 0
    for num in lista:
        if num < digito:
            contador = contador + 1
    return contador

def suma_multiplos_digito(lista, digito):
    suma = 0
    for num in lista:
        if num % digito == 0:
            suma = suma + num
    return suma

def numeros_terminan_con_digito(lista, digito):
    resultado = []
    for num in lista:
        if num % 10 == digito:
            resultado.append(num)
    return resultado

print(f" Cantidad de numeros menores al digito: {menores_al_digito(lista, digito)}")
print(f" Suma de multiplos del digito: {suma_multiplos_digito(lista, digito)}")
print(f" Numeros que terminan con el digito: {numeros_terminan_con_digito(lista, digito)}")
```

## sexto punto


```python
print("SUMADOR DE DIGITOS ")

def suma_digitos(numero):
    numero = abs(numero)
    numero_str = str(numero)
    suma = 0
    for digito in numero_str:
        suma = suma + int(digito)
    return suma

numero = int(input("Ingrese un numero natural: "))

resultado = suma_digitos(numero)

print(f"La suma de los digitos de {numero} es: {resultado}")
```

## septimo punto


```python
print("CONTADOR DE DIGITOS PARES E IMPARES")

def contar_digitos_pares_impares(numero):
    numero = abs(numero)
    numero_str = str(numero)
    pares = 0
    impares = 0
    for digito in numero_str:
        if int(digito) % 2 == 0:
            pares = pares + 1
        else:
            impares = impares + 1
    return pares, impares

numero = int(input("Ingrese un numero natural: "))

pares, impares = contar_digitos_pares_impares(numero)

print(f"El numero {numero} tiene:")
print(f"Digitos pares: {pares}")
print(f"Digitos impares: {impares}")
```
